## 1. Listas

### 1.1 O que são:

- Listas são a estrutura mais usada do Python para guardar vários valores em uma única variável, como uma lista de tarefas, produtos ou nomes de clientes. A documentação oficial define métodos como append, extend, insert, remove, pop e sort para criar, organizar e manipular esses conjuntos de dados.
- Uma lista é uma coleção ordenada de itens. 
- Ela é criada com colchetes [], e os itens ficam separados por vírgulas.

````python
# Uma lista de tarefas para aprender Python
tarefas = [
    "Instalar o Python no Linux",
    "Aprender variáveis",
    "Criar o primeiro programa"
]

print(tarefas)
# Saída: ['Instalar o Python', 'Aprender variáveis', 'Criar o primeiro programa']
````

**Dica**: Pense em uma lista como uma prateleira numerada: cada item ocupa uma posição e pode ser consultado, alterado ou removido.

### 1.2. Caracteristicas


| Característica          | Explicação                                                            | Exemplo                                        |
| ----------------------- | --------------------------------------------------------------------- | ---------------------------------------------- |
| Ordenada                | Os itens mantêm a ordem em que foram colocados.                       | ["a", "b", "c"] é diferente de ["c", "b", "a"] |
| Indexada                | Cada item tem uma posição, começando em 0.                            | tarefas[0] retorna o primeiro item             |
| Mutável                 | Você pode alterar a lista depois de criá-la.                          | tarefas[0] = "Novo valor"                      |
| Aceita repetições       | O mesmo valor pode aparecer mais de uma vez.                          | ["Python", "Python", "Django"]                 |
| Aceita tipos diferentes | Uma lista pode conter textos, números, booleanos e até outras listas. | [1, "Python", True, [2, 3]]                    |

- Exemplo:

````python
# A contagem começa em 0, não em 1
tarefas = ["Instalar o Python", "Aprender variáveis", "Criar o primeiro programa"]

print(tarefas[0])   # Primeiro item: 'Instalar o Python'
print(tarefas[1])   # Segundo item: 'Aprender variáveis'
print(tarefas[-1])  # Último item: 'Criar o primeiro programa'

# Listas são mutáveis: podemos trocar um item existente
tarefas[1] = "Estudar variáveis e tipos"
print(tarefas)
# Saída: ['Instalar o Python', 'Estudar variáveis e tipos', 'Criar o primeiro programa']
````

### 1.3. Quando usar

Use listas quando precisar guardar vários valores relacionados e a ordem deles for importante.
- Lista de produtos em um carrinho de compras.
- Lista de e-mails cadastrados.
- Lista de notas de alunos.
- Lista de posts retornados por uma API.
- Lista de itens selecionados em um formulário Django.

1.3.1 Exemplo:

````python
# Em um sistema Django, uma lista pode representar itens escolhidos
# pelo usuário antes de salvar um pedido no banco de dados.

carrinho = []

# O usuário adiciona produtos ao carrinho
carrinho.append("Teclado mecânico")
carrinho.append("Mouse sem fio")
carrinho.append("Monitor 24 polegadas")

# Antes de salvar, verificamos quantos itens existem
print(f"O carrinho possui {len(carrinho)} produtos.")
# Saída: O carrinho possui 3 produtos.

# Percorrendo a lista para exibir os produtos
for produto in carrinho:
    print(f"- {produto}")
````

- **DICA:** Evite lista quando os itens não podem se repetir: nesse caso, um set costuma ser mais adequado. Use um dicionário quando cada item precisar ter um identificador claro, como "nome" e "preço".


### 1.4. Principais Métodos

#### `append()`

Adiciona **um item** ao final da lista. [docs.python](https://docs.python.org/3.15/tutorial/datastructures.html)

````python
carrinho = ["Teclado mecânico", "Mouse sem fio"]

# Adiciona um novo produto ao final do carrinho
carrinho.append("Monitor 24 polegadas")

print(carrinho)
# Saída: ['Teclado mecânico', 'Mouse sem fio', 'Monitor 24 polegadas']
````

#### `insert()`

Insere um item em uma posição específica, deslocando os demais itens. [docs.python](https://docs.python.org/3.15/tutorial/datastructures.html)

````python
carrinho = ["Teclado mecânico", "Mouse sem fio", "Monitor 24 polegadas"]

# Insere 'Cabo HDMI' na posição 2
# A posição 2 é a terceira posição, porque a contagem começa em 0
carrinho.insert(2, "Cabo HDMI")

print(carrinho)
# Saída: ['Teclado mecânico', 'Mouse sem fio', 'Cabo HDMI', 'Monitor 24 polegadas']
````

#### `extend()`

Adiciona **vários itens** de outra lista ao final da lista atual. [docs.python](https://docs.python.org/3.15/tutorial/datastructures.html)

````python
carrinho = ["Teclado mecânico", "Mouse sem fio"]

# Produtos adicionados em uma promoção
promocao = ["Cabo HDMI", "Suporte para notebook"]

# extend() adiciona cada item separadamente
carrinho.extend(promocao)

print(carrinho)
# Saída: ['Teclado mecânico', 'Mouse sem fio', 'Cabo HDMI', 'Suporte para notebook']
````

**Atenção:** `append(["A", "B"])` adicionaria a lista inteira como um único item; `extend(["A", "B"])` adiciona `"A"` e `"B"` separadamente.

#### `remove()`

Remove a **primeira ocorrência** de um valor específico. [docs.python](https://docs.python.org/3.15/tutorial/datastructures.html)

````python
carrinho = ["Teclado mecânico", "Mouse sem fio", "Cabo HDMI"]

# Remove o produto pelo nome
carrinho.remove("Mouse sem fio")

print(carrinho)
# Saída: ['Teclado mecânico', 'Cabo HDMI']
````

Se o valor não existir, o Python gera um `ValueError`. Uma forma segura é verificar antes:

````python
carrinho = ["Teclado mecânico", "Cabo HDMI"]
produto = "Mouse sem fio"

# Só remove se o produto estiver no carrinho
if produto in carrinho:
    carrinho.remove(produto)
else:
    print(f"O produto '{produto}' não está no carrinho.")
````

#### `pop()`

Remove um item pela **posição** e retorna esse item. Sem índice, remove o último item. [docs.python](https://docs.python.org/3.15/tutorial/datastructures.html)

````python
carrinho = ["Teclado mecânico", "Cabo HDMI", "Monitor 24 polegadas"]

# Remove e guarda o último produto
produto_removido = carrinho.pop()

print(produto_removido)  # Saída: Monitor 24 polegadas
print(carrinho)          # Saída: ['Teclado mecânico', 'Cabo HDMI']

# Remove o item da posição 0
primeiro_produto = carrinho.pop(0)

print(primeiro_produto)  # Saída: Teclado mecânico
print(carrinho)          # Saída: ['Cabo HDMI']
````

#### `index()`

Informa a **posição** da primeira ocorrência de um valor.

````python
carrinho = ["Teclado mecânico", "Cabo HDMI", "Monitor 24 polegadas"]

# Descobre em qual posição está o cabo HDMI
posicao = carrinho.index("Cabo HDMI")

print(posicao)  # Saída: 1
````

#### `count()`

Conta quantas vezes um valor aparece na lista.

````python
carrinho = ["Teclado mecânico", "Cabo HDMI", "Cabo HDMI", "Monitor 24 polegadas"]

# Conta quantas unidades do mesmo produto foram adicionadas
quantidade = carrinho.count("Cabo HDMI")

print(quantidade)  # Saída: 2
````

#### `sort()`

Organiza os itens da própria lista. Em textos, a ordem padrão é alfabética; em números, é crescente. [docs.python](https://docs.python.org/3.15/tutorial/datastructures.html)

````python
carrinho = ["Monitor 24 polegadas", "Cabo HDMI", "Teclado mecânico"]

# Organiza os produtos em ordem alfabética
carrinho.sort()

print(carrinho)
# Saída: ['Cabo HDMI', 'Monitor 24 polegadas', 'Teclado mecânico']

# Organiza em ordem alfabética inversa
carrinho.sort(reverse=True)

print(carrinho)
# Saída: ['Teclado mecânico', 'Monitor 24 polegadas', 'Cabo HDMI']
````

Para ordenar sem alterar a lista original, use a função `sorted()`:

````python
carrinho = ["Monitor 24 polegadas", "Cabo HDMI", "Teclado mecânico"]

# Cria uma nova lista ordenada, mantendo 'carrinho' intacta
carrinho_ordenado = sorted(carrinho)

print(carrinho_ordenado)
# Saída: ['Cabo HDMI', 'Monitor 24 polegadas', 'Teclado mecânico']
print(carrinho)
# Saída: ['Monitor 24 polegadas', 'Cabo HDMI', 'Teclado mecânico']
````

#### `reverse()`

- Inverte a ordem atual dos itens, sem ordená-los alfabeticamente.

````python
carrinho = ["Teclado mecânico", "Cabo HDMI", "Monitor 24 polegadas"]

# Inverte apenas a ordem atual dos produtos
carrinho.reverse()

print(carrinho)
# Saída: ['Monitor 24 polegadas', 'Cabo HDMI', 'Teclado mecânico']
````

#### `copy()`

Cria uma cópia independente da lista.

````python
carrinho = ["Teclado mecânico", "Cabo HDMI"]

# Cria uma cópia para simular uma compra futura
carrinho_futuro = carrinho.copy()

# Alterar a cópia não altera o carrinho original
carrinho_futuro.append("Headset")

print(carrinho)         # Saída: ['Teclado mecânico', 'Cabo HDMI']
print(carrinho_futuro)  # Saída: ['Teclado mecânico', 'Cabo HDMI', 'Headset']
````

#### `clear()`

Remove todos os itens da lista, deixando-a vazia.

````python
carrinho = ["Teclado mecânico", "Cabo HDMI", "Monitor 24 polegadas"]

# Limpa o carrinho após finalizar a compra
carrinho.clear()

print(carrinho)  # Saída: []
````

## Exemplo completo

```python
# Carrinho de compras de uma loja virtual

carrinho = []

# Adicionando produtos
carrinho.append("Teclado mecânico")
carrinho.append("Mouse sem fio")

# Organizando os produtos
carrinho.sort()
print("Carrinho ordenado:", carrinho)

# Finalizando a compra
carrinho.clear()

```

## Resumo rápido

| Método | O que faz | Altera a lista original? |
|---|---|---|
| `append(item)` | Adiciona um item ao final | Sim |
| `insert(posicao, item)` | Insere em uma posição | Sim |
| `extend(outra_lista)` | Adiciona vários itens | Sim |
| `remove(item)` | Remove a primeira ocorrência do valor | Sim |
| `pop()` ou `pop(posicao)` | Remove pela posição e retorna o item | Sim |
| `index(item)` | Retorna a posição do item | Não |
| `count(item)` | Conta ocorrências | Não |
| `sort()` | Ordena a própria lista | Sim |
| `reverse()` | Inverte a ordem | Sim |
| `copy()` | Cria uma cópia | Não |
| `clear()` | Remove todos os itens | Sim |

Esses comportamentos seguem a documentação oficial de estruturas de dados do Python, que descreve `append`, `extend`, `insert`, `remove`, `pop` e `sort` como operações centrais para manipulação de listas. [docs.python](https://docs.python.org/3.15/tutorial/datastructures.html)
