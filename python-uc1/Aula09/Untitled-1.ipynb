## Desafio

- Criar uma lista com 100 números aleatórios
- Calcular a média dos números
- Exibir os os valores maiores que a média


```python
import random
qt_elementos=100
lista=[]
for i in range(qt_elementos):
    lista.append(random.randint(1,999))
media=sum(lista)/len(lista)
print(f"A média dos valores é: {media:.2f}")
maiores=[]
menores=[]
for valor in lista:
    if valor > media :
        maiores.append(valor)
        #print(f"... {valor} ...")
    elif valor < media :
        menores.append(valor)
print (f"{maiores}")
print (f"{menores}")



```
