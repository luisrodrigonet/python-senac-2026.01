 Elabore um programa para construir uma matriz 3x3 usando listas aninhadas e imprima-a. Preencha o conteúdo da matriz com números aleatórios

## Somar duas matrizes usando listas

 Elabore um programa para construir duas matriz 3x3, utilizando listas. A matriz deve ser preenchida por números aleatórios. Ao final imprima o conteúdo de cada matriz, calcule a sua soma e imprima o resultado.

**Código**:

```python
import random  # Biblioteca para gerar números aleatórios

# Função para criar e retornar uma matriz 3x3 com números aleatórios
def criar_matriz():
    matriz = []
    for _ in range(3):  # 3 linhas
        linha = []
        for _ in range(3):  # 3 colunas
            linha.append(random.randint(1, 10))  # Números entre 1 e 10
        matriz.append(linha)
    return matriz

# Função para imprimir uma matriz formatada
def imprimir_matriz(matriz, nome):
    print(f"\n{nome}:")
    for linha in matriz:
        for numero in linha:
            print(f"{numero:3}", end=" ")  # :3 para alinhar os números
        print()  # Nova linha após cada linha da matriz

# Cria as duas matrizes
matriz_A = criar_matriz()
matriz_B = criar_matriz()

# Imprime as matrizes
imprimir_matriz(matriz_A, "Matriz A")
imprimir_matriz(matriz_B, "Matriz B")

# Calcula a soma das matrizes (elemento a elemento)
matriz_soma = []
for i in range(3):  # Percorre as linhas
    linha_soma = []
    for j in range(3):  # Percorre as colunas
        soma = matriz_A[i][j] + matriz_B[i][j]
        linha_soma.append(soma)
    matriz_soma.append(linha_soma)

# Imprime a matriz soma
imprimir_matriz(matriz_soma, "Soma (A + B)")
```

## Desafio : Multiplicar duas matrizes 

 Elabore um programa que utilizando números aleatórios gere duas matrizes 2x2. Em seguida o programa deve multiplique duas matrizes 2x2. O programa deve imprimir as duas matrizes originais e o resultado da multiplicação.

**Código**:

```python
import random

# 1. Gerar duas matrizes 2x2 com números inteiros aleatórios (de 1 a 10)
matriz_a = []
for i in range(2):
    linha = []
    for j in range(2):
        linha.append(random.randint(1, 10))
    matriz_a.append(linha)

matriz_b = []
for i in range(2):
    linha = []
    for j in range(2):
        linha.append(random.randint(1, 10))
    matriz_b.append(linha)

# 2. Inicializar a matriz resultado 2x2 com zeros
resultado = []
for i in range(2):
    linha = []
    for j in range(2):
        linha.append(0)
    resultado.append(linha)

# 3. Multiplicação das matrizes (Linha de A x Coluna de B)
# C[i][j] = A[i][0]*B[0][j] + A[i][1]*B[1][j]
for i in range(2):
    for j in range(2):
        for k in range(2):
            resultado[i][j] += matriz_a[i][k] * matriz_b[k][j]

# 4. Função auxiliar para imprimir matriz de forma formatada
def imprimir_matriz(matriz):
    for linha in matriz:
        for elemento in linha:
            print(f"{elemento:4}", end=" ")
        print()

# 5. Exibição dos resultados
print("Matriz A:")
imprimir_matriz(matriz_a)

print("\nMatriz B:")
imprimir_matriz(matriz_b)

print("\nResultado da Multiplicação (A x B):")
imprimir_matriz(resultado)
```

## Desafio :  Calcular o determinante de uma matriz

 Elabore um programa que utilizando números aleatórios gere duas matrizes 2x2.Calcule o determinante de uma matriz 2x2 usando. O programa deve imprimir as duas matrizes originais e o resultado da operação.

**Código**:

```python
import random

# 1. Gerando a Matriz A (2x2) com números aleatórios
matriz_a = []
for i in range(2):
    linha = []
    for j in range(2):
        linha.append(random.randint(1, 10))
    matriz_a.append(linha)

# 2. Gerando a Matriz B (2x2) com números aleatórios
matriz_b = []
for i in range(2):
    linha = []
    for j in range(2):
        linha.append(random.randint(1, 10))
    matriz_b.append(linha)

# 3. Função para calcular o determinante de uma matriz 2x2
# Fórmula: det = (a00 * a11) - (a01 * a10)
def calcular_determinante_2x2(matriz):
    diagonal_principal = matriz[0][0] * matriz[1][1]
    diagonal_secundaria = matriz[0][1] * matriz[1][0]
    return diagonal_principal - diagonal_secundaria

det_a = calcular_determinante_2x2(matriz_a)
det_b = calcular_determinante_2x2(matriz_b)

# 4. Função auxiliar para imprimir matriz usando for
def imprimir_matriz(matriz):
    for linha in matriz:
        for elemento in linha:
            print(f"{elemento:4}", end=" ")
        print()

# 5. Exibição dos resultados
print("Matriz A:")
imprimir_matriz(matriz_a)
print(f"Determinante de A: {det_a}")

print("\nMatriz B:")
imprimir_matriz(matriz_b)
print(f"Determinante de B: {det_b}")
```

## Desafio : Calcular a soma dos elementos da diagonal principal usando listas

 Elabore um programa que utilizando números aleatórios gere duas matrizes 3x3. Calcule a soma dos elementos da diagonal principal de uma matriz 3x3 usando listas. O programa deve imprimir as duas matrizes originais e o resultado da operação.
 
 **Código**:

```python
import random

# Função para gerar uma matriz 3x3 com números aleatórios
def gerar_matriz():
    matriz = []  # Começa com uma lista vazia
    for i in range(3):  # Para cada linha (3 linhas)
        linha = []      # Cria uma linha vazia
        for j in range(3):  # Para cada coluna (3 colunas)
            # Gera um número aleatório entre 1 e 20 e adiciona à linha
            numero = random.randint(1, 20)
            linha.append(numero)
        matriz.append(linha)  # Adiciona a linha completa à matriz
    return matriz

# Função para calcular a soma da diagonal principal
def soma_diagonal(matriz):
    soma = 0
    for i in range(3):  # Percorre as linhas
        for j in range(3):  # Percorre as colunas
            if i == j:  # Se estamos na diagonal principal
                soma += matriz[i][j]
    return soma

# Programa principal
print("Gerando matriz A...")
matriz_A = gerar_matriz()

print("\nGerando matriz B...")
matriz_B = gerar_matriz()

# Calcula a soma da diagonal de A
resultado = soma_diagonal(matriz_A)

# Mostra as matrizes
print("\nMatriz A:")
for linha in matriz_A:
    print(linha)

print("\nMatriz B:")
for linha in matriz_B:
    print(linha)

# Mostra o resultado
print(f"\nA soma dos elementos da diagonal principal de A é: {resultado}")
```

## Desafio :seven: : Verificar se uma matriz é quadrada

 Elabore um programa que verifique se uma matriz (dada como lista de listas) é quadrada, ou seja, se o número de linhas é igual ao número de colunas.

**Código**:

```python
def verifica_matriz_quadrada(matriz):
    # Verifica se é uma lista
    if type(matriz) != list:
        return False
    
    # Verifica se a matriz está vazia
    if len(matriz) == 0:
        return False
    
    numero_linhas = len(matriz)
    
    # Verifica se todas as linhas são listas
    for linha in matriz:
        if type(linha) != list:
            return False
    
    # Pega o número de colunas da primeira linha
    numero_colunas = len(matriz[0])
    
    # Verifica se todas as linhas têm o mesmo número de colunas
    for linha in matriz:
        if len(linha) != numero_colunas:
            return False
    
    # Verifica se o número de linhas é igual ao de colunas
    if numero_linhas == numero_colunas:
        return True
    else:
        return False

# Exemplo de uso
matriz1 = [[1, 2], [3, 4]]  # Quadrada
matriz2 = [[1, 2, 3], [4, 5, 6]]  # Não quadrada
matriz3 = "isso não é uma matriz"  # Não é matriz
matriz4 = []  # Vazia
matriz5 = [[1], [2], [3]]  # Não quadrada
matriz6 = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]  # Quadrada

# Testando
print("Matriz 1 é quadrada?", verifica_matriz_quadrada(matriz1))
print("Matriz 2 é quadrada?", verifica_matriz_quadrada(matriz2))
print("Matriz 3 é quadrada?", verifica_matriz_quadrada(matriz3))
print("Matriz 4 é quadrada?", verifica_matriz_quadrada(matriz4))
print("Matriz 5 é quadrada?", verifica_matriz_quadrada(matriz5))
print("Matriz 6 é quadrada?", verifica_matriz_quadrada(matriz6))
```

## Desafio : Encontrar o maior valor em uma matriz

 Elabore um programa que crie uma matriz 4×4 preenchida com números aleatórios. O programa deve identificar o maior valor presente na matriz, exibir a matriz gerada e informar qual é o maior valor encontrado.

**Código**:

```python
import random

def criar_matriz():
    """Cria uma matriz 4x4 com números aleatórios entre 1 e 100"""
    matriz = []
    for i in range(4):  # Para cada linha
        linha = []
        for j in range(4):  # Para cada coluna
            linha.append(random.randint(1, 100))
        matriz.append(linha)
    return matriz

def encontrar_maior(matriz):
    """Encontra o maior valor na matriz"""
    maior = matriz[0][0]  # Assume que o primeiro elemento é o maior inicialmente
    for linha in matriz:
        for elemento in linha:
            if elemento > maior:
                maior = elemento
    return maior

def exibir_matriz(matriz):
    """Exibe a matriz formatada"""
    print("\nMatriz gerada:")
    for linha in matriz:
        for elemento in linha:
            print(f"{elemento:4}", end=" ")  # Formata com 4 espaços
        print()  # Nova linha após cada linha da matriz

# Programa principal
matriz = criar_matriz()
maior_valor = encontrar_maior(matriz)

exibir_matriz(matriz)
print(f"\nO maior valor encontrado na matriz é: {maior_valor}")
```

**Funcionamento do programa:**

1. **Geração da matriz**:
   - Cria uma matriz 4×4 com valores inteiros aleatórios entre 1 e 100
   - Usa dois loops `for` aninhados para preencher a matriz

2. **Busca pelo maior valor**:
   - Inicializa a variável `maior` com o primeiro elemento da matriz
   - Compara com todos os outros elementos para encontrar o maior

3. **Exibição dos resultados**:
   - Mostra a matriz completa formatada com alinhamento correto
   - Exibe o maior valor encontrado

**Características importantes:**

- Gera números aleatórios entre 1 e 100 (pode ser ajustado alterando os parâmetros do `random.randint()`)
- Formata a exibição da matriz para melhor visualização
- Identifica corretamente o maior valor independente de sua posição na matriz
- Utiliza estruturas básicas de programação, ideal para aprendizado

## Desafio : Calcular a soma de cada coluna de uma matriz

 Elabore um programa que crie uma matriz 4×4 preenchida com números aleatórios. O programa deve calcular a soma de cada coluna da matriz e exibir o resultado. 

**Código**:

```python
import random

# Criar matriz 4x4 com números aleatórios entre 1 e 10
matriz = []
for i in range(4):
    linha = []
    for j in range(4):
        linha.append(random.randint(1, 10))
    matriz.append(linha)

# Exibir a matriz
print("Matriz:")
for linha in matriz:
    print(linha)

# Calcular a soma de cada coluna
somas_colunas = [0] * 4  # Inicializa uma lista com 4 zeros
for linha in matriz:
    for j in range(4):
        somas_colunas[j] += linha[j]

# Exibir as somas
print("\nSoma das colunas:")
for j, soma in enumerate(somas_colunas):
    print(f"Coluna {j}: {soma}")
```

## Desafio: Verificar se uma matriz contém apenas números positivos

 Elabore um programa que crie uma matriz 4×4 preenchida com números aleatórios, incluindo valores positivos e negativos. O programa deve verificar se todos os números da matriz são positivos ou negativos e contar a quantidade de valores de cada tipo. Por fim, exiba o total de números positivos e negativos encontrados.

**Código**:

```python
import random

# Função para criar matriz 4x4 com números aleatórios positivos e negativos
def criar_matriz():
    matriz = []
    for i in range(4):
        linha = []
        for j in range(4):
            # Gera números entre -50 e 50 (excluindo o zero para este exemplo)
            numero = random.randint(-50, 50)
            while numero == 0:  # Garante que não teremos zeros
                numero = random.randint(-50, 50)
            linha.append(numero)
        matriz.append(linha)
    return matriz

# Função para contar positivos e negativos
def contar_positivos_negativos(matriz):
    positivos = 0
    negativos = 0
    for linha in matriz:
        for numero in linha:
            if numero > 0:
                positivos += 1
            else:
                negativos += 1
    return positivos, negativos

# Função para exibir a matriz
def exibir_matriz(matriz):
    print("\nMatriz gerada:")
    for linha in matriz:
        for numero in linha:
            print(f"{numero:5}", end=" ")  # Formata com 5 espaços
        print()

# Programa principal
matriz = criar_matriz()
exibir_matriz(matriz)

positivos, negativos = contar_positivos_negativos(matriz)

print("\nResultado:")
print(f"Total de números positivos: {positivos}")
print(f"Total de números negativos: {negativos}")

# Verifica se todos são positivos ou negativos
if positivos == 16:
    print("Todos os números na matriz são positivos!")
elif negativos == 16:
    print("Todos os números na matriz são negativos!")
else:
    print("A matriz contém números positivos e negativos.")
```

### Funcionamento do programa:

1. **Geração da matriz**:
   - Cria uma matriz 4×4 com números aleatórios entre -50 e 50
   - Exclui o zero para ter apenas positivos e negativos

2. **Contagem**:
   - Percorre toda a matriz contando números positivos e negativos

3. **Exibição**:
   - Mostra a matriz formatada
   - Exibe a quantidade de números positivos e negativos
   - Informa se todos são do mesmo tipo (todos positivos ou todos negativos)


### Características importantes:
- Gera números aleatórios positivos e negativos
- Formata a exibição para melhor visualização
- Conta corretamente os valores positivos e negativos
- Informa se a matriz contém apenas um tipo de número
- Utiliza estruturas básicas de Python, sem bibliotecas externas

## Desafio : Criar uma matriz com valores aleatórios

 Elabore um programa que crie uma matriz 4×4 preenchida com valores aleatórios entre 0 e 1. Ao final, exiba o conteúdo da matriz. 

```python
import random

# Criar matriz 4x4 com valores aleatórios entre 0 e 1
matriz = []
for i in range(4):
    linha = []
    for j in range(4):
        linha.append(round(random.random(), 4))  # Arredonda para 4 casas decimais
    matriz.append(linha)

# Exibir a matriz formatada
print("Matriz 4x4 (usando listas):")
for linha in matriz:
    for valor in linha:
        print(f"{valor:.4f}", end="  ")  # Formata com 4 casas decimais
    print()  # Nova linha após cada linha da matriz
```
