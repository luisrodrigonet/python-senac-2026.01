## Vetores e Matrizes com Listas em Python

### 1.1 Vetores

Um **vetor** pode ser representado como uma **lista comum** em Python

```python
# Criando um vetor (lista unidimensional)
vetor = [1, 2, 3, 4, 5]
print("Vetor:", vetor)
```

Podemos acessar elementos de um vetor usando **índices**:

```python
print("Primeiro elemento:", vetor[0])  # Saída: 1
print("Último elemento:", vetor[-1])  # Saída: 5
```

### 1.2 Matrizes

Uma matriz pode ser representada como uma **lista de listas**:

```python
# Criando uma matriz 3x3
matriz = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
```

Acessamos elementos usando **dois índices**:

```python
print("Elemento (0,0):", matriz[0][0])  # Saída: 1
print("Elemento (2,1):", matriz[2][1])  # Saída: 8
```

Podemos percorrer matrizes com **loops aninhados**:

```python
for linha in matriz:
    for elemento in linha:
        print(elemento, end=' ')
    print()
```

## 10.1.1 Criando e Exibindo uma Matriz

Elabore um programa que solicite ao usuário para que ele informe, individualmente os valores a serem armazenados em uma **matriz 4×4**. Ao final da leitura dos dados o programa deve exibir a matriz formatada.

1. Utilizando laço de repetição `for`

```python
matriz = []
for i in range(4):
    linha = []
    for j in range(4):
        valor = int(input(f"Digite o valor para [{i}][{j}]: "))
        linha.append(valor)
    matriz.append(linha)

for linha in matriz:
    print(linha)
```

1. Apresentado os dados formatados:

```python
# Inicializa uma matriz 4x4 vazia
matriz = []

print("Digite os valores da matriz 4x4:")

# Loop para ler as 4 linhas
for i in range(4):
    linha = []  # Inicializa uma linha vazia
    
    # Loop para ler os 4 elementos de cada linha
    for j in range(4):
        valor = int(input(f"Digite o valor para a posição [{i}][{j}]: "))
        linha.append(valor)  # Adiciona o valor à linha
    
    matriz.append(linha)  # Adiciona a linha à matriz

# Exibe a matriz formatada
print("\nMatriz 4x4:")
for linha in matriz:
    for elemento in linha:
        print(f"{elemento:4}", end=" ")  # Formata cada elemento com 4 espaços
    print()  # Pula linha após cada linha da matriz
```

📌 **Explicação do Código:**

1. **Inicialização da Matriz:**
   - `matriz = []` → Cria uma lista vazia que armazenará as linhas da matriz.

2. **Preenchimento da Matriz:**
   - O primeiro `for` (`for i in range(4)`) percorre as **linhas** (0 a 3).
   - Dentro dele, `linha = []` cria uma lista vazia para armazenar os elementos da linha atual.
   - O segundo `for` (`for j in range(4)`) percorre as **colunas** (0 a 3).
   - `valor = int(input(...))` solicita ao usuário um número inteiro para a posição `[i][j]`.
   - `linha.append(valor)` adiciona o valor à linha atual.
   - `matriz.append(linha)` adiciona a linha preenchida à matriz principal.

3. **Exibição da Matriz Formatada:**
   - O `for linha in matriz` percorre cada linha da matriz.
   - O `for elemento in linha` percorre cada elemento da linha atual.
   - `print(f"{elemento:4}", end=" ")` formata cada número com 4 espaços para alinhamento.
   - `print()` pula uma linha após cada linha da matriz para manter o formato 4×4.

 **Observações:**
- O código usa `int(input())`, então só aceita números inteiros. Se quiser números decimais, troque para `float(input())`.
- A formatação `{elemento:4}` garante que cada número ocupe 4 espaços, alinhando a matriz corretamente.

## Soma dos Elementos da Diagonal Principal

Elabore um programa que para construir uma **matriz 4×4** preenchida por números aleatórios. Em seguida ele deve calcular a **soma dos elementos da diagonal principal** e realizar a impressão das informações na tela.

```python
import random  # Biblioteca para gerar números aleatórios

# Cria uma matriz 4x4 com números aleatórios entre 1 e 20
matriz = []
for i in range(4):
    linha = []
    for j in range(4):
        linha.append(random.randint(1, 99))  # Números aleatórios de 1 a 99
    matriz.append(linha)

# Exibe a matriz formatada
print("Matriz 4x4 gerada:")
for linha in matriz:
    for elemento in linha:
        print(f"{elemento:4}", end=" ")  # Formatação com 4 espaços
    print()  # Pula linha

# Calcula a soma da diagonal principal
soma_diagonal = 0
for i in range(4):
    soma_diagonal += matriz[i][i]  # Elementos onde linha = coluna (i = j)

# Exibe o resultado
print(f"\nSoma da diagonal principal: {soma_diagonal}")
```

**Explicação do Código:**
1. **Geração da Matriz Aleatória:**
   - `import random` → Importa a biblioteca para gerar números aleatórios.
   - `random.randint(1, 20)` → Gera números inteiros entre 1 e 20.
   - Dois loops `for` aninhados preenchem a matriz 4×4.

2. **Exibição da Matriz:**
   - Dois loops `for` percorrem a matriz e imprimem os valores formatados com `{elemento:4}` para alinhamento.

3. **Cálculo da Diagonal Principal:**
   - A diagonal principal contém os elementos onde o índice da linha é igual ao da coluna (`matriz[i][i]`).
   - Um loop `for` soma esses elementos.

4. **Saída:**
   - Mostra a matriz gerada e a soma da diagonal principal.

**Observações:**
- Se quiser alterar o intervalo dos números aleatórios, basta mudar `random.randint(1, 99)` para outro intervalo (ex: `random.randint(0, 1000)`).
- A diagonal principal é composta pelos elementos onde `i == j` (posições `[0][0]`, `[1][1]`, `[2][2]`, `[3][3]`).

##  Soma dos Elementos da Diagonal Secundária

  Elabore um programa que para construir uma **matriz 4×4** preenchida por números aleatórios. Calcule a **soma dos elementos da diagonal secundária**.

> **DICA:**  A diagonal secundária segue o padrão `matriz[i][3 - i]`.

```python
import random

# Cria uma matriz 4x4 com números aleatórios entre 1 e 100
matriz = []
for i in range(4):
    linha = []
    for j in range(4):
        linha.append(random.randint(1, 100))
    matriz.append(linha)

# Exibe a matriz formatada
print("Matriz 4x4 gerada:")
for linha in matriz:
    for elemento in linha:
        print(f"{elemento:4}", end=" ")
    print()

# Calcula a soma da diagonal secundária
soma_secundaria = 0
for i in range(4):
    soma_secundaria += matriz[i][3 - i]  # Elementos da diagonal secundária

# Exibe o resultado
print(f"\nSoma da diagonal secundária: {soma_secundaria}")
```

**Explicação**:

1. **Geração da Matriz:**
   - Usamos `random.randint(1, 100)` para gerar números inteiros aleatórios entre 1 e 100
   - Dois loops `for` aninhados preenchem a matriz 4×4

2. **Diagonal Secundária:**
   - A diagonal secundária é formada pelos elementos onde a soma dos índices de linha e coluna é igual a 3 (para uma matriz 4×4)
   - Ou seja, as posições são: [0][3], [1][2], [2][1], [3][0]
   - Usamos `matriz[i][3 - i]` para acessar esses elementos

3. **Saída:**
   - Primeiro exibimos a matriz completa formatada
   - Depois mostramos a soma calculada da diagonal secundária

**Observações**:

- Você pode ajustar o intervalo dos números aleatórios alterando os parâmetros em `random.randint()`
- A formatação `{elemento:4}` garante que cada número ocupe 4 espaços para melhor visualização
- A diagonal secundária tem elementos onde i + j = n-1 (para uma matriz n×n)

## Maior Valor de Cada Linha

 Elabore um programa que solicite ao usuário os dados de uma **matriz 4×4**. A leitura dos dados deve ser realizada de forma individual. Ao final o programa deve exibir o **maior elemento de cada linha**.

```python
# Programa que lê uma matriz 4x4 e mostra o maior elemento de cada linha

# Inicializa a matriz vazia
matriz = []

print("Digite os elementos da matriz 4x4:")

# Preenche a matriz com os valores digitados pelo usuário
for i in range(4):
    linha = []
    for j in range(4):
        elemento = int(input(f"Digite o elemento [{i}][{j}]: "))
        linha.append(elemento)
    matriz.append(linha)

# Encontra e exibe o maior elemento de cada linha
print("\nMaior elemento de cada linha:")
for i in range(4):
    maior = max(matriz[i])  # Usa a função max() para encontrar o maior elemento da linha
    print(f"Linha {i}: {maior}")

# Exibe a matriz completa para referência (opcional)
print("\nMatriz completa:")
for linha in matriz:
    for elemento in linha:
        print(f"{elemento:4}", end=" ")
    print()
```

1. **Entrada de dados**:

2. **Processamento**:
   - Para cada linha da matriz, o programa encontra o maior elemento usando a função `max()`
   - O resultado é armazenado e preparado para exibição

3. **Saída**:
   - Mostra o maior elemento de cada linha (linhas 0 a 3)
   - Como bônus, exibe a matriz completa formatada para referência

**Observações:**

- O programa usa `int(input())` para ler números inteiros
- A função `max()` simplifica a busca pelo maior elemento em cada linha
- A exibição da matriz completa é opcional, mas ajuda na verificação dos resultados
- O alinhamento na exibição usa formatação de string (`{elemento:4}`) para melhor visualização

## 5 Contagem de Números Pares

 Elabore um programa que solicite ao usuário que forneça os dados de uma  **matriz 4×4**. Ao final exiba a matriz, a quantidade de números pares e a quantidade de impares ela possui.

```python
# Programa que lê uma matriz 4x4 e conta números pares e ímpares

# Inicializa a matriz e contadores
matriz = []
pares = 0
impares = 0

print("Digite os elementos da matriz 4x4:")

# Preenche a matriz com valores do usuário
for i in range(4):
    linha = []
    for j in range(4):
        while True:
            try:
                elemento = int(input(f"Elemento [{i}][{j}]: "))
                linha.append(elemento)
                
                # Verifica se é par ou ímpar
                if elemento % 2 == 0:
                    pares += 1
                else:
                    impares += 1
                break
            except ValueError:
                print("Por favor, digite um número inteiro válido.")
    matriz.append(linha)

# Exibe a matriz formatada
print("\nMatriz informada:")
for linha in matriz:
    for elemento in linha:
        print(f"{elemento:5}", end=" ")  # Formata com 5 espaços
    print()  # Nova linha

# Exibe os resultados
print(f"\nQuantidade de números pares: {pares}")
print(f"Quantidade de números ímpares: {impares}")
```

**Funcionamento do programa:**
1. **Entrada de dados**:
   - Solicita cada elemento da matriz individualmente
   - Valida se o valor digitado é um número inteiro
   - Classifica cada número como par ou ímpar durante a entrada

2. **Processamento**:
   - Conta os números pares e ímpares enquanto preenche a matriz
   - Armazena os valores em uma estrutura de matriz (lista de listas)

3. **Saída**:
   - Exibe a matriz formatada com alinhamento correto
   - Mostra a contagem de números pares e ímpares

**Melhorias incluídas:**
- Tratamento de erros para garantir que apenas números inteiros sejam aceitos
- Contagem eficiente durante a entrada de dados
- Formatação bonita para exibição da matriz

## Multiplicação de Linhas por um Número

 Elabore um programa que crie uma matriz 4×4 preenchida com números aleatórios. O programa deve, em seguida, exibir o conteúdo da matriz. Após isso, deverá solicitar ao usuário que informe o número de uma linha e um valor inteiro. O programa deverá multiplicar todos os elementos da linha escolhida pelo valor fornecido. Por fim, deve apresentar a matriz atualizada.

```python
import random

# Função para exibir a matriz de forma formatada
def exibir_matriz(matriz, mensagem):
    print(f"\n{mensagem}:")
    for linha in matriz:
        for elemento in linha:
            print(f"{elemento:6}", end=" ")  # Formata com 6 espaços
        print()  # Nova linha

# Cria matriz 4x4 com números aleatórios entre 1 e 20
matriz = [[random.randint(1, 20) for _ in range(4)] for _ in range(4)]

# Exibe a matriz original
exibir_matriz(matriz, "Matriz original")

# Solicita a linha e o multiplicador
while True:
    try:
        linha = int(input("\nDigite o número da linha (0 a 3) para multiplicar: "))
        if 0 <= linha <= 3:
            break
        else:
            print("Por favor, digite um número entre 0 e 3.")
    except ValueError:
        print("Por favor, digite um número inteiro válido.")

while True:
    try:
        multiplicador = int(input("Digite o valor inteiro para multiplicar: "))
        break
    except ValueError:
        print("Por favor, digite um número inteiro válido.")

# Multiplica os elementos da linha escolhida
matriz[linha] = [elemento * multiplicador for elemento in matriz[linha]]

# Exibe a matriz atualizada
exibir_matriz(matriz, "Matriz atualizada")
```

**Funcionamento do programa:**

1. **Geração da Matriz**:

2. **Exibição Inicial**:
   - Mostra a matriz original formatada com números alinhados

3. **Entrada do Usuário**:
   - Solicita o número da linha (0 a 3) com tratamento de erros
   - Pede o valor inteiro multiplicador com validação

4. **Operação de Multiplicação**:
   - Multiplica todos os elementos da linha escolhida pelo valor fornecido
   - Usa list comprehension para uma operação eficiente

5. **Exibição Final**:
   - Mostra a matriz atualizada após a multiplicação

**Características importantes:**

- Validação robusta das entradas do usuário
- Formatação elegante da matriz
- Código limpo e organizado com funções
- Operação vetorizada para multiplicação da linha
- Mensagens claras para o usuário

## Multiplicação de Matrizes

 Elabore um programa que crie duas matrizes 4×4, cada uma preenchida com números aleatórios. O programa deve exibir o conteúdo de ambas as matrizes. Em seguida, calcule a multiplicação matricial entre elas e exiba o resultado.


```python
import random

# Função para criar uma matriz 4x4 com valores aleatórios
def criar_matriz():
    return [[random.randint(1, 10) for _ in range(4)] for _ in range(4)]

# Função para exibir uma matriz
def exibir_matriz(matriz, nome):
    print(f"\n{nome}:")
    for linha in matriz:
        print(" ".join(f"{num:4}" for num in linha))

# Função para multiplicar duas matrizes
def multiplicar_matrizes(matriz_a, matriz_b):
    resultado = [[0 for _ in range(4)] for _ in range(4)]
    for i in range(4):  # linhas da matriz A
        for j in range(4):  # colunas da matriz B
            for k in range(4):  # elementos para multiplicar
                resultado[i][j] += matriz_a[i][k] * matriz_b[k][j]
    return resultado

# Cria as duas matrizes
matriz_a = criar_matriz()
matriz_b = criar_matriz()

# Exibe as matrizes originais
exibir_matriz(matriz_a, "Matriz A")
exibir_matriz(matriz_b, "Matriz B")

# Calcula o produto matricial
matriz_produto = multiplicar_matrizes(matriz_a, matriz_b)

# Exibe o resultado
exibir_matriz(matriz_produto, "Produto matricial A × B")
```

**Funcionamento do programa:**

1. **Geração das Matrizes**:
   - Cria duas matrizes 4×4 com números aleatórios entre 1 e 10
   - Usa list comprehension para criação eficiente

2. **Exibição das Matrizes**:
   - Mostra ambas as matrizes formatadas com alinhamento correto

3. **Multiplicação Matricial**:
   - Implementa o algoritmo clássico de multiplicação de matrizes
   - Triplo loop para calcular cada elemento do produto

4. **Exibição do Resultado**:
   - Mostra a matriz resultante do produto matricial

**Detalhes importantes:**
- O produto matricial não é comutativo (A×B ≠ B×A)
- Cada elemento do resultado é a soma dos produtos dos elementos correspondentes
- O programa usa formatação para alinhar os números corretamente
- O intervalo dos números aleatórios pode ser ajustado modificando o `random.randint(1, 10)`

Para matrizes maiores, seria recomendado usar bibliotecas como NumPy, mas para 4×4 esta implementação direta é eficiente e didática.

## Verificação de Simetria

 Elabore um programa que crie uma matriz 4×4 preenchida com números aleatórios e outra matriz 4×4 simétrica. O programa deve, em seguida, exibir ambas as matrizes. Após isso, ele deve analisar as matrizes e informar se cada uma delas é simétrica, ou seja, se matriz[i][j] == matriz[j][i] para todos os elementos.

```python
import random

def criar_matriz_aleatoria():
    """Cria uma matriz 4x4 com valores aleatórios entre 1 e 9"""
    matriz = []
    for i in range(4):  # Para cada linha
        linha = []
        for j in range(4):  # Para cada coluna
            linha.append(random.randint(1, 9))
        matriz.append(linha)
    return matriz

def criar_matriz_simetrica():
    """Cria uma matriz 4x4 simétrica com valores aleatórios entre 1 e 9"""
    matriz = []
    for i in range(4):
        linha = [0] * 4  # Cria uma linha com 4 zeros
        matriz.append(linha)
    
    for i in range(4):
        for j in range(i, 4):  # Só preenche da diagonal para cima
            valor = random.randint(1, 9)
            matriz[i][j] = valor
            matriz[j][i] = valor  # Garante a simetria
    return matriz

def exibir_matriz(matriz, nome):
    """Exibe uma matriz formatada"""
    print(f"\n{nome}:")
    for linha in matriz:
        for numero in linha:
            print(f"{numero:3}", end=" ")
        print()  # Nova linha após cada linha da matriz

def verificar_simetria(matriz):
    """Verifica se uma matriz é simétrica"""
    for i in range(4):
        for j in range(4):
            if matriz[i][j] != matriz[j][i]:
                return False
    return True

# Programa principal
print("Criação e análise de matrizes 4x4")

# Cria as matrizes
matriz_aleatoria = criar_matriz_aleatoria()
matriz_simetrica = criar_matriz_simetrica()

# Exibe as matrizes
exibir_matriz(matriz_aleatoria, "Matriz Aleatória")
exibir_matriz(matriz_simetrica, "Matriz Simétrica")

# Verifica a simetria
print("\nResultados:")
print("A matriz aleatória é simétrica?", "Sim" if verificar_simetria(matriz_aleatoria) else "Não")
print("A matriz simétrica é simétrica?", "Sim" if verificar_simetria(matriz_simetrica) else "Não")
```

**Funcionamento do programa:**

1. **Criação das Matrizes**:
   - `criar_matriz_aleatoria()`: Gera uma matriz 4×4 com valores aleatórios entre 1 e 9
   - `criar_matriz_simetrica()`: Gera uma matriz simétrica garantindo que matriz[i][j] = matriz[j][i]

2. **Exibição**:
   - As matrizes são exibidas formatadas com números alinhados

3. **Verificação de Simetria**:
   - A função `eh_simetrica()` verifica se todos os elementos obedecem à condição matriz[i][j] == matriz[j][i]


**Pontos importantes:**
- A matriz simétrica é construída garantindo a simetria desde sua criação
- A verificação de simetria compara todos os elementos com seus correspondentes transpostos
- O programa demonstra claramente a diferença entre uma matriz qualquer e uma matriz simétrica

## Normalização de uma Matriz

 Elabore um programa que crie uma matriz 4×4 preenchida com números aleatórios. O programa deve identificar o maior valor presente na matriz e, em seguida, normalizar os demais valores, dividindo-os pelo maior valor encontrado.


```python
import random

def criar_matriz():
    """Cria uma matriz 4x4 com números aleatórios entre 1 e 100"""
    matriz = []
    for i in range(4):
        linha = []
        for j in range(4):
            linha.append(random.randint(1, 100))
        matriz.append(linha)
    return matriz

def encontrar_maior(matriz):
    """Encontra o maior valor na matriz"""
    maior = matriz[0][0]
    for linha in matriz:
        for elemento in linha:
            if elemento > maior:
                maior = elemento
    return maior

def normalizar_matriz(matriz, maior):
    """Normaliza a matriz dividindo cada elemento pelo maior valor"""
    matriz_normalizada = []
    for linha in matriz:
        linha_normalizada = []
        for elemento in linha:
            linha_normalizada.append(round(elemento / maior, 4))  # Arredonda para 4 casas decimais
        matriz_normalizada.append(linha_normalizada)
    return matriz_normalizada

def exibir_matriz(matriz, titulo):
    """Exibe a matriz formatada"""
    print(f"\n{titulo}:")
    for linha in matriz:
        for elemento in linha:
            print(f"{elemento:8.4f}", end=" ")  # Formata com 8 caracteres e 4 decimais
        print()

# Programa principal
matriz_original = criar_matriz()
maior_valor = encontrar_maior(matriz_original)
matriz_normalizada = normalizar_matriz(matriz_original, maior_valor)

exibir_matriz(matriz_original, "Matriz Original")
print(f"\nMaior valor encontrado: {maior_valor}")
exibir_matriz(matriz_normalizada, "Matriz Normalizada")
```

**Funcionamento do programa:**

1. **Criação da matriz**:
   - Gera uma matriz 4×4 com valores inteiros aleatórios entre 1 e 100
   - Usa loops `for` aninhados para construção explícita

2. **Encontrar o maior valor**:
   - Percorre todos os elementos da matriz comparando valores
   - Armazena o maior valor encontrado

3. **Normalização**:
   - Cria uma nova matriz onde cada elemento é dividido pelo maior valor
   - Arredonda os resultados para 4 casas decimais

4. **Exibição**:
   - Mostra a matriz original formatada
   - Exibe o maior valor encontrado
   - Apresenta a matriz normalizada com valores entre 0 e 1


**Observações:**
- Todos os valores da matriz normalizada estarão no intervalo [0, 1]
- O maior valor sempre se tornará 1.0 após a normalização
- O programa usa formatação para exibir os números de forma alinhada
- A normalização preserva a proporção relativa entre os valores originais

##  Busca de um Elemento na Matriz

 Elabore um programa que solicite ao usuário a inserção dos elementos de uma matriz 4×4, lendo-os um a um. Em seguida, o programa deve pedir ao usuário que informe um número. Após isso, exiba o conteúdo da matriz e verifique se o número fornecido está presente, informando sua posição caso seja encontrado.

```python
def criar_matriz():
    """Solicita ao usuário os elementos para preencher uma matriz 4x4"""
    matriz = []
    print("Digite os elementos da matriz 4x4 (linha por linha):")
    for i in range(4):
        linha = []
        for j in range(4):
            while True:
                try:
                    elemento = int(input(f"Elemento [{i}][{j}]: "))
                    linha.append(elemento)
                    break
                except ValueError:
                    print("Por favor, digite um número inteiro válido.")
        matriz.append(linha)
    return matriz

def exibir_matriz(matriz):
    """Exibe a matriz formatada"""
    print("\nMatriz informada:")
    for linha in matriz:
        for elemento in linha:
            print(f"{elemento:5}", end=" ")
        print()

def buscar_numero(matriz, numero):
    """Busca um número na matriz e retorna suas posições"""
    posicoes = []
    for i in range(4):
        for j in range(4):
            if matriz[i][j] == numero:
                posicoes.append((i, j))
    return posicoes

# Programa principal
matriz = criar_matriz()
exibir_matriz(matriz)

numero = int(input("\nDigite um número para buscar na matriz: "))
posicoes = buscar_numero(matriz, numero)

if posicoes:
    print(f"\nO número {numero} foi encontrado nas seguintes posições:")
    for pos in posicoes:
        print(f"Linha: {pos[0]}, Coluna: {pos[1]}")
else:
    print(f"\nO número {numero} não foi encontrado na matriz.")
```

**Funcionamento do programa:**

1. **Entrada de dados**:
   - Solicita ao usuário que digite cada elemento da matriz individualmente
   - Valida se o valor digitado é um número inteiro

2. **Exibição da matriz**:
   - Mostra a matriz formatada com alinhamento correto

3. **Busca do número**:
   - Pede ao usuário um número para buscar na matriz
   - Verifica todas as ocorrências do número na matriz
   - Armazena as posições (linha, coluna) onde o número foi encontrado

4. **Resultado da busca**:
   - Se encontrado, mostra todas as posições onde o número aparece
   - Se não encontrado, informa que o número não está na matriz


**Características importantes:**

- Validação de entrada para garantir que apenas números inteiros sejam aceitos
- Busca completa por todas as ocorrências do número na matriz
- Saída formatada para melhor visualização
- Mensagens claras para o usuário em cada etapa
