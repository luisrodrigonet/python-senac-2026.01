```python
numero_secreto = 7
tentativa1 = int(input("Tente adivinhar o número (1 a 10): "))
tentativa2 = int(input("Tente novamente: "))
tentativa3 = int(input("Última tentativa: "))
if (tentativa1 == numero_secreto) or (tentativa2 == numero_secreto) or (tentativa3 == numero_secreto):
    print("Parabéns! Você acertou!")
else:
    print("Que pena! O número era", numero_secreto)
```

## 7 Crie um programa que:

✅ Peça a idade do usuário.

✅ Se tiver 18 anos ou mais, pode viajar sozinho.

✅ Se tiver 16 ou 17, pode viajar com autorização dos pais.

✅ Se tiver menos de 16, não pode viajar sozinho.

💡 Extra: Use **input()** para perguntar se os pais autorizaram (s/n).

```python
idade = int(input("Digite sua idade: "))
if idade >= 18:
    print("Você pode viajar sozinho!")
elif idade >= 16:
    autorizacao = input("Você tem autorização dos pais? (s/n): ").lower()
    if autorizacao == "s":
        print("Você pode viajar com autorização!")
    else:
        print("Sem autorização, sem viagem!")
else:
    print("Desculpe, você não pode viajar sozinho!")
```

## 🔟 Par ou Ímpar e Positivo ou Negativo? 🔢

Crie um programa que peça um número e informe:

✅ Se ele é par ou ímpar e

✅ Se é positivo ou negativo.

```python
numero = int(input("Digite um número: "))
if numero % 2 == 0:
    paridade = "par"
else:
    paridade = "ímpar"
if numero >= 0:
    sinal = "positivo"
else:
    sinal = "negativo"
print(f"O número {numero} é {paridade} e {sinal}.")
```

## 🔢 Qual Tipo de Triângulo? 🔺

Peça três lados de um triângulo e informe:

1. Equilátero (todos iguais)
2. Isósceles (dois iguais)
5. Escaleno (todos diferentes)

```python
a = float(input("Digite o primeiro lado: "))
b = float(input("Digite o segundo lado: "))
c = float(input("Digite o terceiro lado: "))
if a == b == c:
    print("Triângulo Equilátero")
elif a == b or a == c or b == c:
    print("Triângulo Isósceles")
else:
    print("Triângulo Escaleno")
```

## 🔢 Verificador de Votação 🗳️

Peça a idade e verifique se o usuário pode votar:

1. Obrigatório (18 a 69 anos)
2. Opcional (16, 17 ou 70+)

Proibido (menor de 16)

```python
idade = int(input("Digite sua idade: "))
if 18 <= idade < 70:
    print("Voto obrigatório!")
elif idade == 16 or idade == 17 or idade >= 70:
    print("Voto opcional!️")
else:
    print("Você não pode votar!")
```
