## Desafio 4.1

1️⃣ Peça ao usuário um número e diga se ele é positivo, negativo ou zero.

```python
numero=int(input("Digite um numero inteiro :"))
```

```python
if numero < 0 :
    print(f"O numero .:{numero}:. é NEGATIVO")
elif numero == 0:
    print(f"O numero .:{numero}:. é ZERO")
else :
    print(f"O numero .:{numero}:. é POSITO")
```

## Desafio 4.4

4️⃣ Faça um programa que peça um número e diga se ele é par ou ímpar.

```python
numero=float(input("Digite um numero : "))
```

```python
if (numero == 0 ):
    print (f"O numero .:{numero}:. é ZERO")
elif (numero % 2 == 0) :
    print (f"O numero .:{numero}:. é PAR")
else :
    print (f"O numero .:{numero}:. é IMPAR")
```

## Desafio 4.7

7️⃣ Desenvolva um programa que calcule o IMC de uma pessoa e exiba sua classificação.
Fórmula: IMC = peso / (altura ** 2).

Classificações: 
- Abaixo de 18.5 → Abaixo do peso 
- Entre 18.5 e 24.9 → Peso ideal 
- Entre 25 e 29.9 → Sobrepeso 
- Acima de 30 → Obesidade

```python
peso=float(input("Digite o valor do seu peso (kg)    : "))
altura=float(input("Digite o valor da sua altura (m) : "))
```

```python
peso=50
altura=1.8
imc=peso/(altura ** 2)
```

```python
print (f"O IMC calculado é de {imc}")
if imc < 18.5 :
    print ("Você está abaixo do peso")
elif imc <= 24.9 :
    print ("Você está no peso Ideal")
elif imc <= 29.9 :
    print ("Você está com sobrepeso")
else: 
    print ("Você está com obesidade")

```
