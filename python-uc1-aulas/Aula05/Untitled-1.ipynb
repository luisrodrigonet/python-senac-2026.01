## 8.1 Funçõe simples

```python
def soma (a, b) :
    soma = a + b
    return soma 
```

```python
resultado=soma(12.4,15)
print (f"O resultado é .:{resultado:4.2f}:.")
```

```python
def soma_2 (n1, n2):
    soma=n1+n2
    print(soma)
```

```python
soma_2(12.4,15)
```

```python
resposta=soma_2(12,50)
print(resposta)
```

```python
def potencia (numero, expoente) :
    resultado = numero ** expoente
    return resultado
```

```python
n1=2
exp=8

resultado=potencia(expoente=exp, numero=n1)
print(f"{resultado}")
```

```python
n2=5
exp=4
resultado=potencia(n2,exp)
resultado2=potencia(exp,n2)
```

```python
print (f" {resultado} <> {resultado2}")
```
