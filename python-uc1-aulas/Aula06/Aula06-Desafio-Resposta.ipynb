# 1️⃣ Imprimir números
-🔖 Elabore um código para imprimir os números de 1 a 10 usando for

```python
for numero in range (1, 11):
    print (f"{numero}",end=", ")
```

## 2️⃣ Imprimir os quadrados
- 🔖 Elabore um código para imprimir o quadrado dos números de 1 a 10 usando for

```python
for numero in range (1, 11):
    #quadrado=numero*numero
    quadrado=numero**2
    print (f"O quadrado de .:{numero:02d}:. é .:{quadrado:03d}:.")
```

## 3️⃣ Imprimir Números pares e impares
-🔖 Elabore um código para imprimir todos os números pares e ímpares de 1 a 20 usando for

```python
print (f"Os numero pares de 1 a 20 são: ")
for numero in range (1,21) :
    if (numero % 2 == 0) :
        print (f"{numero:2d}",end=", ")
```

```python
print (f"Os numero impares de 1 a 20 são: ")
for numero in range (1,21) :
    if (numero % 2 == 1) :
        print (f"{numero:2d}",end=", ")
```

```python
pares=[]
impares=[]
for numero in range (1,21) :
    if (numero % 2 == 0) :
        pares.append(numero)
    else :
        impares.append(numero)

print (f"Numeros pares   : {pares}")
print (f"Numeros impares : {impares}")
```

## 4️⃣ Imprimir de 10 até 1
- 🔖 Elabore um código para imprimir de 10 até 1 de forma decrescente, usando while

```python
numero=10
while numero > 0 :
    print (f"{numero}",end=", ")
    numero=numero-1
```

## 6️⃣ Imprimir a soma dos números
- 🔖 Elabore um código para imprimir a soma dos números de 1 a 100 usando for

```python
soma=0
for numero in range (1, 101):
    soma = soma + numero
print (f"O valor da soma de todos os numeros de 1 até 100 é {soma}")
```

```python
soma=0
numero=1
while numero <= 100 :
    soma = soma + numero
    numero = numero + 1
    #print (f"{numero}") 
print (f"A soma dos numeros é {soma}")
```

## 7️⃣ Calcular o fatorial de um número usando while
- 🔖 O programa deve obter um número e em seguida imprimir o valor do fatorial

```python
numero=int(input("Digite um numero :"))
```

```python
numero=10
fatorial=1
numero_aux=numero
while numero_aux >= 1:
    fatorial = fatorial * numero_aux
    numero_aux = numero_aux - 1
print (f"O Fatorial de .:{numero}:. é .:{fatorial}:.")
```
