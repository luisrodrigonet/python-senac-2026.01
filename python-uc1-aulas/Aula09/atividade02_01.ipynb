```python
produtos = (
    ("Arroz", 5.99),
    ("Feijão", 7.49),
    ("Leite", 4.89),
    ("Óleo", 9.99),
    ("Açúcar", 3.29)
)

print("LISTA DE PRODUTOS")
for nome, preco in produtos:
    print(f"{nome:.<20} R$ {preco:6.2f}")
```

```python
produtos = (
    ("Arroz", 5.99, 10),
    ("Feijão", 7.49, 3),
    ("Leite", 4.89, 5),
    ("Óleo", 9.99, 2),
    ("Açúcar", 3.29, 5)
)
total_geral=0
print("LISTA DE PRODUTOS")
for nome, preco, qt in produtos:
    total=preco*qt
    total_geral=total_geral+total
    print(f"{nome:.<20} - {preco:6.2f} x {qt:03} = R$ {total:6.2f}")
total_desconto=total_geral*0.9
print (f"\n\tValor da compra : R$ {total_geral:8.2f}")
print (f"\t       Desconto : - 10% ")
print (f"\tValor    final  : R$ {total_desconto:8.2f}")
```
