### 9.2.1 Cadastro de produtos

**Exercício:**  
Crie um programa que recebe o nome e o preço de **5 produtos**, armazene-os em uma **tupla** e exiba uma listagem formatada.

**Solução:**

```python
produtos = (
    ("Arroz", 5.99),
    ("Feijão", 7.49),
    ("Leite", 4.89),
    ("Óleo", 9.99),
    ("Açúcar", 3.29)
)

print("LISTA DE PRODUTOS")
for nome, preco in produtos:
    print(f"{nome:.<20} R$ {preco:.2f}")
```

```python
produtos = (
    ("Arroz", 5.99, 10),
    ("Feijão", 7.49, 3 ),
    ("Leite", 4.89, 10 ),
    ("Óleo", 9.99, 4),
    ("Açúcar", 3.29, 5)
)

valor_total=0
valor_produto=0

print("\nLISTA DE PRODUTOS\n")
for nome, preco, qt in produtos:
    valor_produto=preco * qt
    #valor_total=valor_total+valor_produto
    valor_total+=valor_produto
    print(f"{nome:.<20} R$ {preco:6.2f} .. QT {qt:03d} .. R$ {valor_produto:8.2f} ")

print (f"\n\tValor Total da Compra: R$ {valor_total:7.2f}")
```

### 9.2.3 Verificação de números primos

**Exercício:**  
Crie uma **tupla** com **10 números inteiros** e exiba apenas os **números primos**.

**Solução:**

```python
numeros = (2, 7, 10, 13, 17, 22, 29, 31, 40, 53)

def eh_primo(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

primos_lista = []
for num in numeros:
    if eh_primo(num):
        primos_lista.append(num)

print(f"Números primos: {primos_lista}")
primos = tuple(primos_lista)
print(f"Números primos: {primos}")
```

### 9.2.4 Classificação de idade

**Exercício:**  
Peça ao usuário **5 idades** e exiba quantas pessoas são **maiores de idade**.

**Solução:**

```python
import random

#idades = (18, 25, 16, 30, 14)

idades=[]
for _ in range (100):
    idades.append(random.randint(1,92))

#print(len(idades))

maiores = 0
for idade in idades:
    if idade >= 18:
        maiores += 1

print(f"Quantidade de maiores de idade: {maiores}")
```

### 9.2.10 Jogo de dados

 **Exercício:**  
Simule o lançamento de um **dado** 5 vezes e exiba quantas vezes cada número foi sorteado.

**Solução:**

```python
import random

# Gerando os 5 lançamentos com for tradicional
lancamentos_lista = []
for _ in range(5):
    lancamentos_lista.append(random.randint(1, 6))

lancamentos = tuple(lancamentos_lista)
print(f"Resultados: {lancamentos}")

# Contagem tradicional
for num in range(1, 7):
    print(f"{num} saiu {lancamentos.count(num)} vezes.")
```
