```python
vetor = [1, 2, 3, 4, 5]
multiplicador=2
```

```python
vetor_mult = [elemento * multiplicador for elemento in vetor]
```

```python
print(vetor_mult)
```

```python
vetor_mult=[]
for elemento in vetor :
    vetor_mult.append( elemento * multiplicador)
```

```python
print (vetor_mult)
```

```python
produtos = (
    ("Arroz", 5.99),
    ("Feijão", 7.49),
    ("Leite", 4.89),
    ("Óleo", 9.99),
    ("Açúcar", 3.29)
)

print("LISTA DE PRODUTOS")
for nome, preco in produtos:
    print(f"{nome:_<30} R$ {preco:.2f}")
```
