```python
#Exemplo de Vetor
vetor=[10, 20, 30, 40, 50]
print(f" Dados do Vetor: {vetor}")

print(f" Dados da 4a posicao : {vetor[3]}")
```

```python
for elemento in vetor :
    print (f"Valor do Elemento {elemento}")
```

- O valor do 1o elemento é 10
- O valor do 2o elemento é 20

```python
for i in range (5) :
    print (f"{i}")
    print (f"{vetor[i]}")    
```

```python
for i in range (5) :
    print (f"O valor da {i+1}a posicao é {vetor[i]}")
```

```python
indice = 1
for elemento in vetor :
    print (f"O valor do indice {indice} é {elemento}")
    indice+=1
```

```python
import random
numero = random.randint(1,187)
print(f"{numero}")
```

```python
numeros=[]
for i in range (20) :
    numeros.append(random.randint(1,187))

#print(f"{numeros}")

soma=sum(numeros)
maior=max(numeros)
menor=min(numeros)
qt_elementos=len(numeros)
media=soma/qt_elementos
```

```python
print(f"A soma dos elementos é  : {soma}")
print(f"A soma dos elementos é  : {soma:6}")
print(f"A soma dos elementos é  : {soma:09}")
print(f"O maior valor é         : {maior}")
print(f"O menor valor é         : {menor}")
print(f"A qt de elementos é     : {qt_elementos}")
print(f"O valor médio é         : {media:6}")
```
