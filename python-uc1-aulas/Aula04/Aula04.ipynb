## Estrutura Condicional - Desvio Condicional

```python
nome=input("Digite o seu nome : ")
ano_nasc=int(input(f"{nome}, digite o ano do seu nascimento : "))
```

```python
idade=2026-ano_nasc
#idade=12
```

```python
if idade >= 18 :
    print (f"{nome}, você é MAIOR de idade !")
elif idade >= 16:
    print (f"{nome}, você é Adolecente de idade !")
else :
    print (f"{nome}, você é menor de idade !")
```
