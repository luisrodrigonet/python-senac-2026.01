```python
def soma(a, b):
   return a + b

def subtracao(a, b):
   return a - b

def main():
   resultado_soma = soma(10, 5)
   resultado_subtracao = subtracao(10, 5)
   print(f"Soma: {resultado_soma}")
   print(f"Subtração: {resultado_subtracao}")

```

```python
if __name__ == "__main__":
   main()
```
