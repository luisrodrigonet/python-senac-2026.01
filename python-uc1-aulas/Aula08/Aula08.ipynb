## Aula 08 - Funções

### Passagem de argumento com valor "Default"

```python
def saudacao(display="sem nome") :
    print (f"Olá, {display} !!!")
```

```python
saudacao("Papagaio")
```

```python
saudacao()
```

## Passagem de argumentos arbitrario

```python
def saudacoes (*nomes) :
    #print (nomes)
    for nome in nomes :
        saudacao(nome)
```

```python
saudacoes ("Luis", "Rodrigo", "Papagaio")
```

## Testando se o código está sendo carregado como modulo ou como programa

```python
def saudacao():
   print("Bem-vindo ao Python!")
```

```python
def saudacao2():
   print("Bem-vindo ao Python!")
```

```python
def main():
    saudacao()
    saudacao2()
```

```python
if __name__ == "__main__" :
    main()
```
