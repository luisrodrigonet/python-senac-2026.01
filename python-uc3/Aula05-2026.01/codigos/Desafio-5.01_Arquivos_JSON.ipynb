1. Bibliotecas

```python
# Módulo para tratar o arquivo JSON
import json
# Módulo para verificar a existência do arquivo
import os 
```

2. Variaveis

```python
# Nome do Arquivo
ARQUIVO_AGENDA = "agenda.json"
```

3. Carregar dados

```python
def carregar_contatos():
    """
    Carrega a lista de contatos do arquivo JSON.
    Cria o arquivo ou retorna uma lista vazia se ele não existir ou estiver vazio.
    """
    if not os.path.exists(ARQUIVO_AGENDA):
        # 1. Tratamento de Arquivo Inexistente
        print(f"Arquivo '{ARQUIVO_AGENDA}' não encontrado. Criando nova agenda vazia.")
        return []

    try:
        with open(ARQUIVO_AGENDA, 'r', encoding='utf-8') as f:
            # 2. Leitura com tratamento de arquivo vazio (Modo 'r')
            conteudo = f.read()
            if not conteudo:
                return []
            
            # 3. Deserialização (JSON para Objeto Python)
            return json.loads(conteudo)
            
    except json.JSONDecodeError:        
        # Tratamento para JSON malformado 
        print(f"AVISO: O arquivo '{ARQUIVO_AGENDA}' está malformado. Iniciando com uma lista vazia.")
        return []

    except Exception as e:
        print(f"Erro inesperado ao carregar contatos: {e}")
        return []
```

4. Salvar Dados

```python
def salvar_contatos(contatos):
    """
    Salva a lista completa de contatos no arquivo JSON.
    """
    try:
        # 4. Escrita no arquivo (Modo 'w' - sobrescreve)
        with open(ARQUIVO_AGENDA, 'w', encoding='utf-8') as f:
            # Serialização (Objeto Python para JSON)
            # 'indent=4' é uma boa prática para tornar o JSON legível por humanos.
            json.dump(contatos, f, indent=4, ensure_ascii=False)
        print("\n Agenda salva com sucesso!")
    except Exception as e:
        print(f"\n ERRO ao salvar a agenda: {e}")
```

5. Adiciona novo Contato

```python
"""
Adiciona um novo contato à lista e salva no arquivo.
"""
def adicionar_contato(contatos):
    
    print("\n--- Adicionar Novo Contato ---")
    nome = input("Nome: ").strip()
    telefone = input("Telefone: ").strip()
    email = input("Email: ").strip()

    if not nome or not telefone:
        print("Nome e Telefone são campos obrigatórios. Operação cancelada.")
        return

    novo_contato = {
        "nome": nome,
        "telefone": telefone,
        "email": email
    }
    
    # Adiciona o novo contato a lista
    contatos.append(novo_contato)

    # Salva as informações sobre o contato no arquivo
    salvar_contatos(contatos)
```

6. Listar Contatos

```python
"""
Exibe todos os contatos de forma formatada.
"""
def listar_contatos(contatos):
    
    print("\n--- Lista de Contatos ---")
    if not contatos:
        print("A agenda está vazia.")
        return

    for i, contato in enumerate(contatos):
        print("-" * 30)
        # Utilizando f-strings para uma formatação clara
        print(f"ID: {i+1}") 
        print(f"Nome: {contato.get('nome', 'N/A')}")
        print(f"Telefone: {contato.get('telefone', 'N/A')}")
        print(f"Email: {contato.get('email', 'N/A')}")
    print("-" * 30)
    print(f"Total de contatos: {len(contatos)}")
```

7. Buscar Contatos

```python
"""
Busca contatos pelo nome (parcial ou completo).
"""
def buscar_contato(contatos):
    
    if not contatos:
        print("A agenda está vazia. Nada para buscar.")
        return
        
    termo = input("\nDigite o nome (ou parte do nome) para buscar: ").strip().lower()
    
    # Filtragem: compara o termo de busca (minúsculo) com o nome do contato (minúsculo)
    #encontrados = [c for c in contatos if termo in c.get('nome', '').lower()]
    encontrados = []
    for c in contatos:
        # Garantimos que 'nome' existe e convertemos para minúsculas para a busca case-insensitive
        nome_contato = c.get('nome', '').lower() 
        
        if termo in nome_contato:
            encontrados.append(c)

    if not encontrados:
        print(f"Nenhum contato encontrado com o termo '{termo}'.")
        return

    print(f"\n--- Contatos Encontrados (Total: {len(encontrados)}) ---")
    for i, contato in enumerate(encontrados):
        print(f"Nome: {contato['nome']}, Telefone: {contato['telefone']}")
    print("-" * 30)   
```

8. Excluir um contato

```python
"""
Exclui um contato da lista pelo nome.
"""
def excluir_contato(contatos: list):
    
    # Exibe a lista para o usuário ver os IDs
    listar_contatos(contatos) 
    if not contatos:
        return

    try:
        # Solicita o ID (índice + 1) para facilitar a exclusão
        id_excluir = int(input("Digite o ID do contato a ser EXCLUÍDO (ou 0 para cancelar): "))
        
        if id_excluir == 0:
            print("Exclusão cancelada.")
            return

        # Ajusta o ID para o índice real da lista (ID - 1)
        indice = id_excluir - 1 

        if 0 <= indice < len(contatos):
            # Remove o item pelo índice
            contato_removido = contatos.pop(indice) 

            # Salva a nova lista sem o contato
            salvar_contatos(contatos) 
            print(f"Contato '{contato_removido['nome']}' removido com sucesso!")
        else:
            print("ID inválido. Por favor, digite um número da lista.")

    except ValueError:
        print("Entrada inválida. Digite um número.")
    except Exception as e:
        print(f"Erro ao tentar excluir: {e}")
```

9. Menu Principal

```python
"""
Exibe o menu de opções e controla o fluxo do programa.
"""
def menu_principal():
    
    # Carrega os dados ao iniciar o programa (executa o tratamento de arquivo)
    agenda_contatos = carregar_contatos() 

    while True:
        print("\n*** Agenda de Contatos em JSON ***")
        print("1. Adicionar Contato")
        print("2. Listar Contatos")
        print("3. Buscar Contato")
        print("4. Excluir Contato")
        print("5. Sair")
        
        escolha = input("Escolha uma opção: ").strip()

        if escolha == '1':
            adicionar_contato(agenda_contatos)
        elif escolha == '2':
            listar_contatos(agenda_contatos)
        elif escolha == '3':
            buscar_contato(agenda_contatos)
        elif escolha == '4':
            excluir_contato(agenda_contatos)
        elif escolha == '5':
            print("Saindo da Agenda. Até logo!")
            break
        else:
            print("Opção inválida. Tente novamente.")

# Bloco de execução principal
if __name__ == "__main__":
    menu_principal()
```
