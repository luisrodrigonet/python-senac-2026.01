- Bibliotecas

```python
import csv
import os
```

- Variaveis

```python
ARQUIVO_AGENDA = "agenda.csv"
CAMPOS = ["nome", "telefone", "email"] 
```

- Carregar Dados

```python
def carregar_contatos():

    if not os.path.exists(ARQUIVO_AGENDA):
        print(f"Arquivo '{ARQUIVO_AGENDA}' não encontrado. Criando nova agenda vazia.")
        return []

    try:
        with open(ARQUIVO_AGENDA, 'r', newline='', encoding='utf-8') as f:
            reader = csv.DictReader(f)
            return list(reader)
            
    except Exception as e:
        print(f"Erro inesperado ao carregar contatos do CSV: {e}")
        return []
```

- Salvar Dados

```python
def salvar_contatos(contatos):

    try:        
        with open(ARQUIVO_AGENDA, 'w', newline='', encoding='utf-8') as f:
            writer = csv.DictWriter(f, fieldnames=CAMPOS)
            writer.writeheader()
            writer.writerows(contatos)            
        print("\nAgenda salva com sucesso no formato CSV!")

    except Exception as e:
        print(f"\nERRO ao salvar a agenda no CSV: {e}")
```

- Adicionar Contato

```python
def adicionar_contato(contatos):

    print("\n--- Adicionar Novo Contato ---")
    nome = input("Nome: ").strip()
    telefone = input("Telefone: ").strip()
    email = input("Email: ").strip()

    if not nome or not telefone:
        print("Nome e Telefone são campos obrigatórios. Operação cancelada.")
        return

    novo_contato = {
        "nome": nome,
        "telefone": telefone,
        "email": email
    }
    
    contatos.append(novo_contato)
    salvar_contatos(contatos)
```

- Listar dados

```python
def listar_contatos(contatos):

    print("\n--- Lista de Contatos ---")
    if not contatos:
        print("A agenda está vazia.")
        return

    for i, contato in enumerate(contatos):
        print("-" * 30)
        print(f"ID: {i+1}") 
        print(f"Nome: {contato.get('nome', 'N/A')}")
        print(f"Telefone: {contato.get('telefone', 'N/A')}")
        print(f"Email: {contato.get('email', 'N/A')}")
    print("-" * 30)
    print(f"Total de contatos: {len(contatos)}")
```

```python
contatos=carregar_contatos()
```

```python
print(contatos)
```

```python
adicionar_contato(contatos)
```

```python
listar_contatos(contatos)
```
